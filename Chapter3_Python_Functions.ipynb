# 1️⃣ METHODS (PHƯƠNG THỨC)

## 📖 Theory

**Method** là một hàm được **gắn liền với một đối tượng (object)** cụ thể, gọi thông qua cú pháp `object.method()`. Khác với **Function** (hàm độc lập, gọi trực tiếp bằng tên), Method luôn "thuộc về" một kiểu dữ liệu nào đó (string, list, dict...).

### So sánh Method và Function

| Đặc điểm | Function | Method |
|---|---|---|
| Cách gọi | `ten_ham(tham_so)` | `object.ten_method(tham_so)` |
| Gắn với đối tượng? | Không | **Có** — luôn thuộc về 1 object |
| Ví dụ | `print("Hi")`, `len(x)` | `"Hi".upper()`, `list.append(x)` |

### Một số Method phổ biến theo kiểu dữ liệu

```
   str  → .upper()  .lower()  .strip()  .split()  .replace()
   list → .append() .remove() .sort()   .reverse() .pop()
   dict → .keys()   .values() .items()  .get()
```

## 💻 Example — Ví dụ thực tế


````python
# Method của kiểu String - xử lý dữ liệu văn bản "bẩn" (thực tế hay gặp khi làm sạch dữ liệu)
email_input = "  NguyenVanA@GMAIL.com  "

email_sach = email_input.strip().lower()   # xóa khoảng trắng thừa + chuyển chữ thường
print("Email trước khi xử lý:", repr(email_input))
print("Email sau khi xử lý  :", repr(email_sach))

# Tách họ tên thành list các từ
ho_ten = "Nguyễn Văn An"
cac_tu = ho_ten.split(" ")
print("Các từ trong họ tên:", cac_tu)
````

````python
# Method của kiểu List - quản lý danh sách công việc (to-do list)
todo = ["Học Python", "Làm bài tập"]

todo.append("Ôn thi giữa kỳ")     # thêm việc mới
print("Sau khi thêm:", todo)

todo.sort()                        # sắp xếp theo alphabet
print("Sau khi sắp xếp:", todo)

# Method của kiểu Dictionary - lấy dữ liệu an toàn với .get()
sinh_vien = {"ten": "Lan", "tuoi": 19}
print(sinh_vien.get("ten"))              # "Lan"
print(sinh_vien.get("truong", "Chưa có"))  # key không tồn tại -> trả về giá trị mặc định
````

## 📊 Visualization

```
   "hello".upper()
        │
        ▼
   ┌─────────────────────────────┐
   │  Object: "hello" (string)    │
   │  .upper()  ── method gọi ──►  "HELLO"  │
   └─────────────────────────────┘

   Method LUÔN đi kèm dấu chấm "." sau một object cụ thể:
        object   .   method_name   ( tham_so )
        "hello"  .   upper         (        )

# 2️⃣ FUNCTIONS (HÀM)

## 📖 Theory

**Function (Hàm)** là một khối lệnh được **đặt tên**, có thể **tái sử dụng** nhiều lần mà không cần viết lại code.

### Cú pháp khai báo hàm

```python
def ten_ham(tham_so_1, tham_so_2):
    # khối lệnh xử lý
    return ket_qua       # (tùy chọn) trả về giá trị
```

### Các thành phần của một hàm

| Thành phần | Ý nghĩa |
|---|---|
| `def` | Từ khóa bắt buộc để khai báo hàm |
| `ten_ham` | Tên hàm (nên đặt có ý nghĩa) |
| `(tham_so)` | Tham số đầu vào (parameters) — có thể không có |
| `return` | Trả kết quả ra ngoài (nếu không có `return`, hàm trả về `None`) |

### Vì sao cần dùng Hàm?

```
❌ KHÔNG dùng hàm (lặp code nhiều lần):
   dtb1 = (8 + 7 + 9) / 3
   dtb2 = (5 + 6 + 7) / 3
   dtb3 = (9 + 9 + 10) / 3

✅ Dùng hàm (viết 1 lần, dùng nhiều lần):
   def tinh_trung_binh(a, b, c):
       return (a + b + c) / 3

   dtb1 = tinh_trung_binh(8, 7, 9)
   dtb2 = tinh_trung_binh(5, 6, 7)
   dtb3 = tinh_trung_binh(9, 9, 10)
```

## 💻 Example — Ví dụ thực tế


````python
# Hàm tính điểm trung bình 3 môn - tái sử dụng cho nhiều sinh viên
def tinh_diem_trung_binh(toan, ly, hoa):
    # Tính điểm trung bình của 3 môn học
    dtb = (toan + ly + hoa) / 3
    return round(dtb, 2)

# Gọi hàm nhiều lần với dữ liệu khác nhau
print("Sinh viên A:", tinh_diem_trung_binh(8, 7, 9))
print("Sinh viên B:", tinh_diem_trung_binh(5, 6, 7))
````

````python
# Hàm với tham số mặc định (default parameter) - ví dụ tính tiền hóa đơn có thuế
def tinh_tien_hoa_don(gia_goc, thue_suat=0.1):
    # Tính tổng tiền sau thuế. Mặc định thuế suất 10%
    tien_thue = gia_goc * thue_suat
    return gia_goc + tien_thue

print("Hóa đơn 100.000đ (thuế mặc định 10%):", tinh_tien_hoa_don(100000))
print("Hóa đơn 100.000đ (thuế 8%):", tinh_tien_hoa_don(100000, 0.08))
````

# 3️⃣ SCOPE (PHẠM VI BIẾN)

## 📖 Theory

**Scope** xác định **nơi một biến có thể được truy cập** trong chương trình. Python có 2 phạm vi chính:

| Loại Scope | Định nghĩa | Truy cập được ở đâu? |
|---|---|---|
| **Local** | Biến khai báo **bên trong hàm** | Chỉ bên trong hàm đó |
| **Global** | Biến khai báo **bên ngoài** mọi hàm | Toàn bộ chương trình |

### Quy tắc quan trọng

- Biến **local** chỉ tồn tại trong lúc hàm đang chạy, mất đi khi hàm kết thúc
- Hàm **có thể đọc** biến global, nhưng **không thể sửa** nó trực tiếp (trừ khi dùng từ khóa `global`)
- Nên **hạn chế** dùng `global` — dễ gây lỗi khó kiểm soát khi code lớn

## 📊 Visualization

```
   ┌──────────────────────────────────────────────┐
   │              GLOBAL SCOPE                      │
   │   diem_chuan = 5      ← biến global             │
   │                                                 │
   │   def kiem_tra(diem):                           │
   │   ┌────────────────────────────────────┐        │
   │   │         LOCAL SCOPE (trong hàm)      │        │
   │   │   diem       ← tham số (local)       │        │
   │   │   ket_qua = diem >= diem_chuan        │        │
   │   │   (hàm ĐỌC được diem_chuan từ ngoài)  │        │
   │   │   return ket_qua                       │        │
   │   └────────────────────────────────────┘        │
   │                                                 │
   │   print(ket_qua)  ❌ LỖI! ket_qua là biến local,│
   │                     không tồn tại ở ngoài hàm    │
   └──────────────────────────────────────────────┘
```

## 💻 Example — Ví dụ thực tế


````python
diem_chuan = 5   # biến GLOBAL - áp dụng chung cho cả hệ thống

def kiem_tra_dau(diem):
    # 'diem' và 'ket_qua' là biến LOCAL, chỉ tồn tại trong hàm này
    ket_qua = "Đậu" if diem >= diem_chuan else "Rớt"
    return ket_qua

print(kiem_tra_dau(6))   # hàm vẫn đọc được diem_chuan (global)
print(kiem_tra_dau(3))

# Thử truy cập biến local từ bên ngoài -> sẽ báo lỗi nếu bỏ comment dòng dưới
# print(ket_qua)   # NameError: name 'ket_qua' is not defined
````

````python
# Minh họa: hàm KHÔNG thể thay đổi biến global nếu không dùng từ khóa "global"
so_du = 1000000   # biến global - số dư tài khoản

def rut_tien_sai(so_tien):
    so_du = so_du - so_tien   # ❌ Lỗi! Python coi đây là biến LOCAL mới, chưa được gán giá trị
    return so_du

# rut_tien_sai(200000)   # Nếu bỏ comment sẽ báo lỗi: UnboundLocalError

def rut_tien_dung(so_tien):
    global so_du              # ✅ khai báo rõ: dùng biến GLOBAL "so_du"
    so_du = so_du - so_tien
    return so_du

print("Số dư sau khi rút:", rut_tien_dung(200000))
print("Biến global so_du đã thay đổi:", so_du)
````

- Hàm **đọc** được biến global tự nhiên, nhưng cần từ khóa `global` để **thay đổi** nó
- Nên hạn chế `global` — ưu tiên dùng `return` để trả kết quả rõ ràng hơn

# 4️⃣ LAMBDA (HÀM ẨN DANH)

## 📖 Theory

**Lambda** là cách viết một hàm **ngắn gọn, không cần đặt tên** (anonymous function), thường dùng cho các phép tính đơn giản, dùng 1 lần.

### Cú pháp

```python
lambda tham_so: bieu_thuc
```

### So sánh Function thường và Lambda

```
   Function thường:                    Lambda tương đương:

   def binh_phuong(x):                 binh_phuong = lambda x: x ** 2
       return x ** 2
```

⚠️ Lambda chỉ chứa **1 biểu thức duy nhất**, không có `if/else` phức tạp hay nhiều dòng lệnh như hàm thường (trừ dùng biểu thức điều kiện ngắn gọn).

## 💻 Example — Ví dụ thực tế


````python
# Lambda cơ bản
binh_phuong = lambda x: x ** 2
print(binh_phuong(5))    # 25

# Lambda với nhiều tham số
tinh_tong = lambda a, b: a + b
print(tinh_tong(3, 4))   # 7

# Lambda với điều kiện (dùng if-else dạng biểu thức ngắn gọn)
phan_loai = lambda diem: "Đậu" if diem >= 5 else "Rớt"
print(phan_loai(7))
print(phan_loai(3))
````

````python
# Ví dụ thực tế: dùng lambda làm "key" để sắp xếp dữ liệu phức tạp
san_pham = [
    {"ten": "Áo thun", "gia": 150000},
    {"ten": "Quần jean", "gia": 350000},
    {"ten": "Mũ lưỡi trai", "gia": 80000},
]

# Sắp xếp danh sách sản phẩm theo giá tăng dần
san_pham_sap_xep = sorted(san_pham, key=lambda sp: sp["gia"])

for sp in san_pham_sap_xep:
    print(sp["ten"], "-", sp["gia"], "đ")
````

## 📊 Visualization

```
   Function thường (nhiều dòng, có tên):
   ┌───────────────────────────┐
   │ def binh_phuong(x):         │
   │     return x ** 2            │
   └───────────────────────────┘

   Lambda (1 dòng, không tên - "vô danh"):
   ┌───────────────────────────┐
   │  lambda x: x ** 2            │───► gán vào biến để dùng lại,
   └───────────────────────────┘      hoặc dùng trực tiếp làm tham số
                                        cho sorted(), map(), filter()...

# 5️⃣ MAP()

## 📖 Theory

`map()` áp dụng **một hàm** lên **từng phần tử** của một dãy dữ liệu (list, tuple...), trả về kết quả tương ứng cho mỗi phần tử — thay thế cho vòng lặp `for` khi cần "biến đổi" toàn bộ dữ liệu.

### Cú pháp

```python
map(ham, day_du_lieu)
```

⚠️ `map()` trả về một **đối tượng map** (map object) — cần bọc `list()` để xem kết quả dạng danh sách.

### So sánh: vòng lặp `for` vs `map()`

```
❌ Dùng for (dài dòng hơn):
   gia_moi = []
   for gia in gia_list:
       gia_moi.append(gia * 1.1)

✅ Dùng map() (ngắn gọn hơn):
   gia_moi = list(map(lambda gia: gia * 1.1, gia_list))
```

## 💻 Example — Ví dụ thực tế


````python
# Ví dụ: tăng giá sản phẩm thêm 10% cho toàn bộ danh sách
gia_goc = [100000, 250000, 75000, 500000]

gia_moi = list(map(lambda gia: gia * 1.1, gia_goc))
print("Giá gốc:", gia_goc)
print("Giá sau khi tăng 10%:", gia_moi)
````

````python
# map() kết hợp với function thường (không chỉ lambda)
def chuyen_do_C_sang_F(do_C):
    return do_C * 9/5 + 32

nhiet_do_C = [0, 20, 30, 37, 100]
nhiet_do_F = list(map(chuyen_do_C_sang_F, nhiet_do_C))
print("Độ C:", nhiet_do_C)
print("Độ F:", nhiet_do_F)
````

## 📊 Visualization

```

# 6️⃣ FILTER()

## 📖 Theory

`filter()` dùng để **lọc ra** các phần tử **thỏa mãn điều kiện** (hàm trả về `True`) trong một dãy dữ liệu — khác với `map()` (biến đổi giá trị), `filter()` chỉ **giữ lại hoặc loại bỏ** phần tử.

### Cú pháp

```python
filter(ham_dieu_kien, day_du_lieu)
```

⚠️ Cũng giống `map()`, cần bọc `list()` để xem kết quả.

### So sánh `map()` và `filter()`

| Hàm | Chức năng | Số lượng phần tử đầu ra |
|---|---|---|
| `map()` | **Biến đổi** từng phần tử | Bằng với đầu vào |
| `filter()` | **Lọc/giữ lại** phần tử thỏa điều kiện | Có thể **ít hơn** đầu vào |

## 💻 Example — Ví dụ thực tế


````python
# Ví dụ: lọc ra danh sách sinh viên đạt học bổng (điểm >= 8.5)
diem_list = [8.7, 6.5, 9.0, 5.5, 8.9, 7.2]

diem_hoc_bong = list(filter(lambda d: d >= 8.5, diem_list))
print("Điểm đạt học bổng:", diem_hoc_bong)
````

````python
# Kết hợp map() và filter(): lọc số chẵn rồi nhân đôi từng số
so_list = list(range(1, 21))

so_chan = list(filter(lambda x: x % 2 == 0, so_list))
print("Số chẵn:", so_chan)

so_chan_nhan_2 = list(map(lambda x: x * 2, so_chan))
print("Số chẵn sau khi nhân 2:", so_chan_nhan_2)
````

# 7️⃣ *ARGS (THAM SỐ KHÔNG GIỚI HẠN)

## 📖 Theory

`*args` cho phép một hàm nhận **số lượng tham số không giới hạn** (không biết trước bao nhiêu), Python sẽ gom chúng lại thành một **tuple**.

### Cú pháp

```python
def ten_ham(*args):
    # args là một tuple chứa mọi tham số được truyền vào
```

> 💡 Tên `args` chỉ là quy ước phổ biến, dấu `*` mới là quan trọng — bạn có thể đặt tên khác như `*so`, `*values`.

## 💻 Example — Ví dụ thực tế


````python
# Ví dụ: hàm tính tổng doanh thu không giới hạn số lượng cửa hàng
def tinh_tong_doanh_thu(*doanh_thu):
    print("Kiểu dữ liệu của doanh_thu:", type(doanh_thu))
    return sum(doanh_thu)

# Gọi hàm với số lượng tham số khác nhau - đều hợp lệ!
print("Tổng 2 cửa hàng:", tinh_tong_doanh_thu(50000000, 70000000))
print("Tổng 4 cửa hàng:", tinh_tong_doanh_thu(50000000, 70000000, 30000000, 45000000))
````

````python
# Ví dụ thực tế 2: hàm tìm điểm cao nhất trong số lượng môn học bất kỳ
def tim_diem_cao_nhat(*diem):
    if len(diem) == 0:
        return None
    return max(diem)

print("Điểm cao nhất:", tim_diem_cao_nhat(8, 6, 9, 7))
print("Điểm cao nhất:", tim_diem_cao_nhat(5.5, 9.5))
````

# 8️⃣ **KWARGS (THAM SỐ TỪ KHÓA KHÔNG GIỚI HẠN)

## 📖 Theory

`**kwargs` (keyword arguments) cho phép hàm nhận **số lượng tham số dạng `key=value` không giới hạn**, Python gom lại thành một **dictionary**.

### Cú pháp

```python
def ten_ham(**kwargs):
    # kwargs là 1 dictionary chứa các cặp key=value được truyền vào
```

### So sánh `*args` và `**kwargs`

| Đặc điểm | `*args` | `**kwargs` |
|---|---|---|
| Kiểu dữ liệu bên trong hàm | `tuple` | `dict` |
| Cách gọi hàm | `ham(1, 2, 3)` | `ham(a=1, b=2, c=3)` |
| Truy cập | Theo vị trí (`args[0]`) | Theo tên key (`kwargs["a"]`) |

## 💻 Example — Ví dụ thực tế


````python
# Ví dụ: tạo hồ sơ sinh viên linh hoạt (không biết trước sẽ có bao nhiêu thông tin)
def tao_ho_so(**thong_tin):
    print("Kiểu dữ liệu của thong_tin:", type(thong_tin))
    for key, value in thong_tin.items():
        print(f"  {key}: {value}")

tao_ho_so(ten="Lan", tuoi=19, nganh="Data Science")
print("---")
tao_ho_so(ten="Nam", tuoi=20, nganh="CNTT", que_quan="Cần Thơ", so_thich="Bóng đá")
````

````python
# Kết hợp *args và **kwargs trong cùng 1 hàm
def in_don_hang(ma_don, *san_pham, **thong_tin_khach):
    print("Mã đơn hàng:", ma_don)
    print("Sản phẩm:", san_pham)
    print("Thông tin khách hàng:")
    for key, value in thong_tin_khach.items():
        print(f"  {key}: {value}")

in_don_hang("DH001", "Áo thun", "Quần jean", ten="Lan", dia_chi="Cần Thơ", sdt="0909xxxxxx")
````

# 9️⃣ PRACTICE TỔNG HỢP (ÔN TẬP CHƯƠNG 3)

## 📖 Theory

Phần này kết hợp: **Methods, Functions, Scope, Lambda, Map, Filter, \*args, \*\*kwargs**.

## 💻 Example — Bài toán mẫu

**Đề bài:** Xây dựng hệ thống phân tích điểm số linh hoạt cho lớp học, có thể nhận số lượng sinh viên bất kỳ.


````python
diem_chuan_gioi = 8.5   # biến global

def phan_tich_lop(**sinh_vien_diem):
    # Nhận vào số lượng sinh viên bất kỳ dạng ten=diem,
    # trả về danh sách sinh viên giỏi và điểm trung bình lớp.
    danh_sach_diem = list(sinh_vien_diem.values())
    trung_binh_lop = sum(danh_sach_diem) / len(danh_sach_diem)

    # Dùng filter + lambda để lọc sinh viên giỏi
    sinh_vien_gioi = list(filter(lambda item: item[1] >= diem_chuan_gioi, sinh_vien_diem.items()))

    return round(trung_binh_lop, 2), sinh_vien_gioi

diem_tb, ds_gioi = phan_tich_lop(An=8.7, Binh=6.5, Chi=9.0, Dung=5.5, Em=8.9)

print("Điểm trung bình lớp:", diem_tb)
print("Danh sách sinh viên giỏi:")
for ten, diem in ds_gioi:
    print(f"  {ten}: {diem}")
````

# 🔟 GITHUB: BRANCH, MERGE, PULL REQUEST, CONFLICT

## 📖 Theory

Ở Chương 2, bạn đã học quy trình cơ bản: `clone → commit → push → pull`. Khi làm việc **theo nhóm**, cần thêm các khái niệm sau để nhiều người có thể cùng code mà **không giẫm chân nhau**.

### 1. Branch (Nhánh)

**Branch** là một "nhánh rẽ" độc lập của code, cho phép bạn phát triển tính năng mới **mà không ảnh hưởng** đến nhánh chính (`main`).

```bash
git branch tinh-nang-moi        # tạo nhánh mới
git checkout tinh-nang-moi       # chuyển sang nhánh đó
# hoặc gộp 2 lệnh trên:
git checkout -b tinh-nang-moi
```

### 2. Merge (Gộp nhánh)

**Merge** là quá trình **gộp thay đổi** từ một nhánh vào nhánh khác (thường là gộp nhánh tính năng vào `main`).

```bash
git checkout main
git merge tinh-nang-moi
```

### 3. Pull Request (PR)

**Pull Request** là một **yêu cầu chính thức** trên GitHub để gộp nhánh của bạn vào nhánh chính, cho phép đồng đội **review code** trước khi merge — cực kỳ quan trọng khi làm việc nhóm hoặc đóng góp mã nguồn mở.

```
1. Push nhánh của bạn lên GitHub
2. Vào GitHub -> "Compare & pull request"
3. Mô tả thay đổi -> Tạo Pull Request
4. Đồng đội review, bình luận, yêu cầu sửa (nếu cần)
5. Khi được duyệt -> "Merge pull request"
```

### 4. Conflict (Xung đột)

**Conflict** xảy ra khi Git **không thể tự động gộp** 2 thay đổi vì chúng sửa **cùng một dòng code** theo cách khác nhau — cần con người **quyết định giữ phiên bản nào**.

## 📊 Visualization — Sơ đồ Branch Workflow

```
main:        A───B───C───────────────F───G          (nhánh chính)
                      \                 /
feature/xyz:           D───E──────────/              (nhánh tính năng)

    A,B,C = các commit trên main (trước khi tạo branch)
    D,E   = các commit trên nhánh feature/xyz (làm tính năng mới)
    F     = commit merge (gộp feature/xyz vào main qua Pull Request)
    G     = tiếp tục phát triển trên main sau khi merge
```

### Quy trình làm việc nhóm chuẩn với Pull Request

```
┌─────────────────────────────────────────────────────────┐
│ 1. git checkout -b feature/them-bieu-do                  │
│    → Tạo nhánh riêng cho tính năng "thêm biểu đồ"          │
├─────────────────────────────────────────────────────────┤
│ 2. Code, git add, git commit trên nhánh này                │
├─────────────────────────────────────────────────────────┤
│ 3. git push origin feature/them-bieu-do                    │
│    → Đẩy nhánh lên GitHub                                   │
├─────────────────────────────────────────────────────────┤
│ 4. Tạo Pull Request trên GitHub: feature → main             │
├─────────────────────────────────────────────────────────┤
│ 5. Đồng đội Review code, comment góp ý                      │
├─────────────────────────────────────────────────────────┤
│ 6. Sửa theo góp ý (nếu có) → push lại → PR tự cập nhật       │
├─────────────────────────────────────────────────────────┤
│ 7. Merge Pull Request → code chính thức vào main            │
└─────────────────────────────────────────────────────────┘
```

### Minh họa Conflict xảy ra như thế nào

```
   File: bao_cao.py

   Trên nhánh main (bạn A sửa):        Trên nhánh feature (bạn B sửa):
   ────────────────────────            ─────────────────────────
   ten_du_an = "Dự án Data A"           ten_du_an = "Dự án Data B"

                    │                              │
                    └──────────────┬───────────────┘
                                    ▼
                         git merge feature vào main
                                    │
                                    ▼
                    ⚠️ CONFLICT! Git không biết giữ dòng nào

   <<<<<<< HEAD
   ten_du_an = "Dự án Data A"
   =======
   ten_du_an = "Dự án Data B"
   >>>>>>> feature

   👉 Con người phải MỞ FILE, chọn giữ dòng nào (hoặc gộp cả 2),
      xóa các ký hiệu <<<<<<< ======= >>>>>>>, rồi commit lại.
```

### Cách giải quyết Conflict — các bước

```bash
# 1. Git báo conflict khi merge/pull
git merge feature/xyz
# Auto-merging bao_cao.py

````python
# Mô phỏng khái niệm Conflict: 2 người cùng sửa "cùng 1 dòng" theo cách khác nhau
ban_A_sua = {"dong_1": "Dự án Data A"}
ban_B_sua = {"dong_1": "Dự án Data B"}

if ban_A_sua["dong_1"] != ban_B_sua["dong_1"]:
    print("⚠️  CONFLICT phát hiện tại 'dong_1':")
    print("   <<<<<<< HEAD")
    print("  ", ban_A_sua["dong_1"])
    print("   =======")
    print("  ", ban_B_sua["dong_1"])
    print("   >>>>>>> feature-branch")
    print("👉 Cần con người quyết định giữ nội dung nào rồi commit lại.")
````

## 📌 Summary

- **Branch**: nhánh code độc lập, giúp phát triển tính năng mới an toàn, không ảnh hưởng `main`
- **Merge**: gộp thay đổi từ 1 nhánh vào nhánh khác
- **Pull Request**: yêu cầu chính thức để gộp code, cho phép **review** trước khi merge
- **Conflict**: xảy ra khi 2 thay đổi "đụng độ" trên cùng 1 dòng code, cần con người xử lý thủ công
- Quy trình chuẩn khi làm nhóm: `branch mới → code/commit → push → Pull Request → review → merge`
