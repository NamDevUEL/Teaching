# 1️⃣ CÂU LỆNH `if`

## 📖 Theory

Câu lệnh `if` cho phép chương trình **ra quyết định**: chỉ thực thi một khối lệnh khi điều kiện là `True`.

### Cú pháp

```python
if dieu_kien:
    # khối lệnh thực thi khi dieu_kien = True
```

⚠️ **Lưu ý quan trọng:**
- Sau `if dieu_kien` phải có dấu **hai chấm `:`**
- Khối lệnh bên trong phải được **thụt lề (indent)** — thường là 4 dấu cách

### Sơ đồ luồng (Flowchart)

```
              ┌───────────────┐
              │  Điều kiện?    │
              └───────┬───────┘
                       │
             ┌─────────┴─────────┐
             ▼ True               ▼ False
      ┌─────────────┐      (không làm gì,
      │ Thực thi     │       bỏ qua khối if)
      │ khối lệnh    │
      └─────────────┘
```

## 💻 Example


````python
# Ví dụ cơ bản về if
tuoi = 20

if tuoi >= 18:
    print("Bạn đã đủ tuổi trưởng thành")
````

````python
# if với nhiều điều kiện kết hợp (and/or)
diem = 8.5
chuyen_can = 90

if diem >= 8 and chuyen_can >= 80:
    print("Bạn đủ điều kiện nhận học bổng!")
````

# 2️⃣ CÂU LỆNH `elif`

## 📖 Theory

`elif` (viết tắt của "else if") cho phép kiểm tra **thêm điều kiện khác** nếu điều kiện `if` phía trên là `False`. Có thể có **nhiều `elif`**.

### Cú pháp

```python
if dieu_kien_1:
    # thực thi nếu dieu_kien_1 True
elif dieu_kien_2:
    # thực thi nếu dieu_kien_1 False và dieu_kien_2 True
elif dieu_kien_3:
    # thực thi nếu 2 điều kiện trên False và dieu_kien_3 True
```

### Sơ đồ luồng

```
        if dk1? ──True──► thực thi khối if
          │False
          ▼
       elif dk2? ──True──► thực thi khối elif
          │False
          ▼
       elif dk3? ──True──► thực thi khối elif
          │False
          ▼
     (không khối nào chạy nếu không có else)
```

## 💻 Example


````python
# Xếp loại học lực dựa trên điểm số
diem = 7.5

if diem >= 9:
    print("Xếp loại: Xuất sắc")
elif diem >= 8:
    print("Xếp loại: Giỏi")
elif diem >= 6.5:
    print("Xếp loại: Khá")
elif diem >= 5:
    print("Xếp loại: Trung bình")
else:
    print("Xếp loại: Yếu")
````

# 3️⃣ CÂU LỆNH `else`

## 📖 Theory

`else` là khối lệnh được thực thi khi **tất cả** điều kiện `if`/`elif` phía trên đều `False`. `else` **không có điều kiện** đi kèm.

### Cú pháp đầy đủ if - elif - else

```python
if dieu_kien_1:
    # ...
elif dieu_kien_2:
    # ...
else:
    # thực thi khi tất cả điều kiện trên đều False
```

### Sơ đồ luồng đầy đủ

```
        ┌──────────┐
        │  if dk1? │──True──► Khối if
        └────┬─────┘
             │False
        ┌────▼─────┐
        │ elif dk2?│──True──► Khối elif
        └────┬─────┘
             │False
        ┌────▼─────┐
        │   else   │────────► Khối else (luôn chạy nếu
        └──────────┘           không điều kiện nào đúng)
```

## 💻 Example


````python
# Kiểm tra số chẵn/lẻ
so = 17

if so % 2 == 0:
    print(so, "là số chẵn")
else:
    print(so, "là số lẻ")
````

````python
# Kết hợp if - elif - else đầy đủ: kiểm tra đăng nhập
username = "admin"
password = "12345"

if username == "admin" and password == "12345":
    print("Đăng nhập thành công!")
elif username == "admin" and password != "12345":
    print("Sai mật khẩu!")
else:
    print("Tài khoản không tồn tại!")
````

# 4️⃣ VÒNG LẶP `for`

## 📖 Theory

Vòng lặp `for` dùng để **lặp qua từng phần tử** của một chuỗi (list, tuple, string, range...) và thực thi một khối lệnh cho mỗi phần tử.

### Cú pháp

```python
for phan_tu in day_du_lieu:
    # khối lệnh thực thi cho mỗi phan_tu
```

### Sơ đồ minh họa

```
   day_du_lieu = [10, 20, 30]

   Vòng lặp 1: phan_tu = 10 ──► thực thi khối lệnh
   Vòng lặp 2: phan_tu = 20 ──► thực thi khối lệnh
   Vòng lặp 3: phan_tu = 30 ──► thực thi khối lệnh
   Hết phần tử ──► kết thúc vòng lặp
```

## 💻 Example


````python
# Lặp qua từng phần tử của List
trai_cay = ["táo", "cam", "chuối"]

for tc in trai_cay:
    print("Tôi thích ăn", tc)
````

````python
# Lặp qua từng ký tự của chuỗi (string)
ten = "Lan"
for ky_tu in ten:
    print(ky_tu)

# Tính tổng các phần tử trong List bằng for
diem = [8, 7, 9, 6, 10]
tong = 0
for d in diem:
    tong = tong + d
print("Tổng điểm:", tong)
````

# 5️⃣ VÒNG LẶP `while`

## 📖 Theory

Vòng lặp `while` lặp lại một khối lệnh **chừng nào điều kiện còn đúng (`True`)**. Khác với `for` (lặp theo số phần tử biết trước), `while` phù hợp khi **không biết trước số lần lặp**.

### Cú pháp

```python
while dieu_kien:
    # khối lệnh lặp lại khi dieu_kien = True
```

⚠️ **Cảnh báo Vòng lặp vô hạn (Infinite Loop):** Nếu điều kiện **không bao giờ** trở thành `False`, chương trình sẽ chạy **mãi mãi**! Luôn nhớ **cập nhật biến điều kiện** bên trong vòng lặp.

### Sơ đồ luồng

```
        ┌───────────────┐
   ┌───►│  Điều kiện?    │
   │    └───────┬───────┘
   │            │True
   │      ┌─────▼──────┐
   │      │ Thực thi    │
   │      │ khối lệnh   │
   │      └─────┬──────┘
   └────────────┘
                │False
                ▼
          Thoát vòng lặp
```

## 💻 Example


````python
# Đếm từ 1 đến 5 bằng while
dem = 1
while dem <= 5:
    print("Đếm:", dem)
    dem = dem + 1   # QUAN TRỌNG: cập nhật biến để tránh lặp vô hạn

print("Đã đếm xong!")
````

````python
# Ví dụ thực tế: rút tiền cho đến khi hết số dư
so_du = 100
rut = 30

while so_du >= rut:
    so_du -= rut
    print("Đã rút", rut, "- Số dư còn lại:", so_du)

print("Không đủ tiền để rút thêm!")
````

- ⚠️ Luôn nhớ cập nhật biến điều kiện để tránh **vòng lặp vô hạn**

# 6️⃣ CÂU LỆNH `break`

## 📖 Theory

`break` dùng để **thoát ngay lập tức** khỏi vòng lặp (`for` hoặc `while`), bất kể điều kiện lặp còn đúng hay không.

### Sơ đồ minh họa

```
   for i in [1, 2, 3, 4, 5]:
        nếu i == 3: break  ──►  Thoát khỏi vòng lặp ngay lập tức

   Kết quả in ra:  1, 2   (dừng lại khi gặp break, không in 3, 4, 5)
```

## 💻 Example


````python
# break trong vòng lặp for
for so in range(1, 10):
    if so == 5:
        print("Gặp số 5, dừng vòng lặp!")
        break
    print("Số hiện tại:", so)
````

````python
# Ví dụ thực tế: tìm số đầu tiên chia hết cho 7 trong danh sách
danh_sach = [3, 8, 12, 14, 21, 25]

for so in danh_sach:
    if so % 7 == 0:
        print("Tìm thấy số chia hết cho 7:", so)
        break
else:
    print("Không tìm thấy số nào chia hết cho 7")
````

# 7️⃣ CÂU LỆNH `continue`

## 📖 Theory

`continue` dùng để **bỏ qua phần còn lại** của lần lặp hiện tại và **nhảy sang lần lặp kế tiếp** — khác với `break` (thoát hẳn vòng lặp).

### So sánh `break` và `continue`

```
   for i in [1, 2, 3, 4, 5]:

   break tại i=3:      1, 2                (dừng hẳn)
   continue tại i=3:   1, 2, 4, 5           (bỏ qua 3, tiếp tục các số sau)
```

## 💻 Example


````python
# continue: bỏ qua số chẵn, chỉ in số lẻ
for so in range(1, 11):
    if so % 2 == 0:
        continue     # bỏ qua, nhảy sang vòng lặp kế tiếp
    print(so)
````

````python
# Ví dụ thực tế: bỏ qua giá trị không hợp lệ khi xử lý danh sách điểm
diem_list = [8, -1, 7, 15, 9, -5]   # -1, 15, -5 là điểm không hợp lệ (phải từ 0-10)

diem_hop_le = []
for d in diem_list:
    if d < 0 or d > 10:
        continue     # bỏ qua điểm không hợp lệ
    diem_hop_le.append(d)

print("Danh sách điểm hợp lệ:", diem_hop_le)
````

## 📌 Summary

- Khác với `break`: `continue` **không thoát** khỏi vòng lặp
- Hữu ích khi cần **lọc bỏ** các trường hợp không mong muốn trong vòng lặp

# 8️⃣ HÀM `range()`

## 📖 Theory

`range()` tạo ra một **dãy số nguyên**, thường dùng kết hợp với vòng lặp `for` khi cần lặp theo số lần cụ thể (thay vì lặp qua phần tử có sẵn).

### 3 cách dùng `range()`

| Cú pháp | Ý nghĩa | Ví dụ | Kết quả |
|---|---|---|---|
| `range(stop)` | Từ 0 đến `stop - 1` | `range(5)` | `0,1,2,3,4` |
| `range(start, stop)` | Từ `start` đến `stop - 1` | `range(2, 6)` | `2,3,4,5` |
| `range(start, stop, step)` | Từ `start` đến `stop-1`, bước nhảy `step` | `range(0, 10, 2)` | `0,2,4,6,8` |

### Sơ đồ minh họa `range(2, 10, 2)`

```
   start=2                          stop=10 (không bao gồm)
     │                                  │
     ▼                                  ▼
     2 ──►  4 ──►  6 ──►  8    (bước nhảy = 2, dừng trước 10)
```

## 💻 Example


````python
# range(stop)
for i in range(5):
    print(i, end=" ")
print()

# range(start, stop)
for i in range(2, 8):
    print(i, end=" ")
print()

# range(start, stop, step)
for i in range(0, 20, 5):
    print(i, end=" ")
print()

# range với step âm - đếm ngược
for i in range(10, 0, -2):
    print(i, end=" ")
````

# 9️⃣ HÀM `enumerate()`

## 📖 Theory

`enumerate()` giúp lặp qua một dãy dữ liệu (list, tuple...) mà **vừa lấy được index, vừa lấy được giá trị** cùng lúc — tiện hơn nhiều so với việc tự quản lý biến đếm.

### Cú pháp

```python
for index, gia_tri in enumerate(day_du_lieu):
    # sử dụng cả index lẫn gia_tri
```

### So sánh: KHÔNG dùng vs CÓ dùng `enumerate()`

```
❌ Cách "thủ công" (dễ sai sót):
   i = 0
   for tc in trai_cay:
       print(i, tc)
       i += 1

✅ Cách dùng enumerate() (ngắn gọn, an toàn):
   for i, tc in enumerate(trai_cay):
       print(i, tc)
```

### `enumerate()` có thể tùy chỉnh điểm bắt đầu

```python
enumerate(day_du_lieu, start=1)   # index bắt đầu từ 1 thay vì 0
```

## 💻 Example

# 🔟 NUMPY

## 📖 Theory

**NumPy** (Numerical Python) là thư viện nền tảng cho tính toán khoa học trong Python, cung cấp cấu trúc dữ liệu **mảng (array)** mạnh mẽ và **nhanh hơn nhiều** so với List thông thường.

### Vì sao dùng NumPy thay vì List?

| Tiêu chí | List | NumPy Array |
|---|---|---|
| Tốc độ tính toán | Chậm | **Nhanh hơn nhiều lần** |
| Phép toán trên toàn bộ mảng | Phải dùng vòng lặp | Tính trực tiếp (vectorized) |
| Dùng cho Data Science | Hạn chế | **Tiêu chuẩn** (nền tảng của Pandas) |

### Cài đặt NumPy

```bash
pip install numpy
```

### Import NumPy

```python
import numpy as np    # "np" là quy ước viết tắt phổ biến
```

### Sơ đồ minh họa Array 1 chiều và 2 chiều

```
   Mảng 1 chiều (1D):     [10, 20, 30, 40]

   Mảng 2 chiều (2D):     [[1, 2, 3],
                            [4, 5, 6]]     ← giống 1 bảng (matrix) 2 dòng, 3 cột
```

## 💻 Example


````python
import numpy as np

# Tạo mảng NumPy từ list
a = np.array([1, 2, 3, 4, 5])
print("Mảng a:", a)
print("Kiểu dữ liệu:", type(a))
print("Shape (kích thước):", a.shape)
````

````python
# So sánh List và NumPy khi tính toán trên toàn bộ phần tử

# Cách dùng List (phải dùng vòng lặp)
list_thuong = [1, 2, 3, 4, 5]
list_nhan_2 = []
for x in list_thuong:
    list_nhan_2.append(x * 2)
print("List nhân 2 (thủ công):", list_nhan_2)

# Cách dùng NumPy (vectorized - không cần vòng lặp!)
arr = np.array([1, 2, 3, 4, 5])
arr_nhan_2 = arr * 2
print("NumPy array nhân 2:", arr_nhan_2)
````

````python
# Các hàm thống kê phổ biến của NumPy
diem = np.array([8, 7.5, 9, 6, 10, 5.5])

print("Tổng:", np.sum(diem))
print("Trung bình:", np.mean(diem))
print("Lớn nhất:", np.max(diem))
print("Nhỏ nhất:", np.min(diem))
print("Độ lệch chuẩn:", round(np.std(diem), 2))

# Tạo mảng đặc biệt
print("Mảng số 0:", np.zeros(5))
print("Mảng số 1:", np.ones(4))
print("Mảng dãy số:", np.arange(0, 10, 2))    # giống range() nhưng trả về array
````

````python
# Mảng 2 chiều (ma trận) - hữu ích khi biểu diễn bảng dữ liệu
diem_lop = np.array([
    [8, 7, 9],     # điểm sinh viên 1: Toán, Lý, Python
    [6, 8, 7],     # điểm sinh viên 2
    [9, 9, 10]     # điểm sinh viên 3
])

print("Ma trận điểm:\n", diem_lop)
print("Shape:", diem_lop.shape)   # (3 dòng, 3 cột)
print("Điểm trung bình từng môn:", np.mean(diem_lop, axis=0))   # theo cột
print("Điểm trung bình từng sinh viên:", np.mean(diem_lop, axis=1))  # theo dòng
````

````python
# Lời giải Bài 12.1: Tìm số nguyên tố từ 2 đến 50
print("Các số nguyên tố từ 2 đến 50:")
for so in range(2, 51):
    la_nguyen_to = True
    for i in range(2, so):
        if so % i == 0:
            la_nguyen_to = False
            break
    if la_nguyen_to:
        print(so, end=" ")
````

# 1️⃣3️⃣ GITHUB WORKFLOW

## 📖 Theory

Ở Chương 1, bạn đã biết cách **tạo Repository và Upload file** trực tiếp trên web. Trong chương này, ta tìm hiểu **quy trình làm việc chuẩn (Workflow)** với Git — dùng dòng lệnh (command line), quy trình mà mọi lập trình viên/Data Scientist đều dùng hàng ngày.

### 4 khái niệm cốt lõi: Clone – Commit – Push – Pull

| Lệnh | Ý nghĩa | Khi nào dùng |
|---|---|---|
| `git clone` | **Sao chép** một repository từ GitHub về máy | Khi bắt đầu làm việc với 1 dự án có sẵn |
| `git commit` | **Lưu lại** một điểm mốc thay đổi của code (kèm mô tả) | Sau khi hoàn thành 1 phần việc |
| `git push` | **Đẩy** các commit từ máy tính lên GitHub | Khi muốn chia sẻ thay đổi lên cloud |
| `git pull` | **Kéo** các thay đổi mới nhất từ GitHub về máy | Khi đồng đội đã cập nhật code trên GitHub |

### Sơ đồ luồng dữ liệu: Working Directory → Staging → Commit → GitHub

```
┌───────────────────┐      git add        ┌───────────────────┐
│  Working Directory  │ ──────────────────► │      Staging       │
│  (Thư mục làm việc) │                     │   (Vùng chuẩn bị)   │
│                     │                     │                     │
│  Nơi bạn chỉnh sửa  │                     │  Các file đã được   │
│  file trực tiếp     │                     │  đánh dấu sẽ lưu    │
└───────────────────┘                     └──────────┬────────┘
                                                       │
                                                       │ git commit -m "..."
                                                       ▼
                                            ┌───────────────────┐
                                            │       Commit        │
                                            │  (Lịch sử local)    │
                                            │                     │
                                            │  Một "điểm mốc" đã  │
                                            │  lưu vĩnh viễn vào  │
                                            │  lịch sử Git local  │
                                            └──────────┬────────┘
                                                       │
                                                       │ git push
                                                       ▼
                                            ┌───────────────────┐
                                            │       GitHub         │
                                            │   (Lưu trữ Cloud)    │
                                            │                     │
                                            │  Nơi lưu trữ code    │
                                            │  trên internet,      │
                                            │  chia sẻ với mọi người│
                                            └───────────────────┘
```

### Chu trình làm việc điển hình hàng ngày

```
   1. git clone <link>      → Tải project về máy (chỉ làm 1 lần đầu)
            │
            ▼
   2. Chỉnh sửa code trong Working Directory
            │
            ▼
   3. git add <file>        → Đưa file vào Staging
            │
            ▼
   4. git commit -m "..."   → Lưu lại thay đổi (Commit)
            │
            ▼
   5. git push               → Đẩy lên GitHub
            │
            ▼
   6. git pull               → (Trước khi làm tiếp) Lấy code mới nhất
            │
            ▼
      Quay lại bước 2, lặp lại vòng lặp làm việc mỗi ngày 🔁
```

### Chi tiết từng lệnh

**1. `git clone` — Sao chép Repository về máy**
```bash
git clone https://github.com/username/ten-repo.git
```
→ Tạo ra một thư mục mới trên máy, chứa toàn bộ code + lịch sử của repository đó.

**2. `git commit` — Lưu lại thay đổi**
```bash
git add ten_file.py          # bước 1: đưa file vào staging
git commit -m "Mô tả thay đổi"  # bước 2: lưu lại (commit)
```
→ Mỗi commit giống như "chụp ảnh" trạng thái code tại một thời điểm, kèm mô tả rõ ràng.

**3. `git push` — Đẩy code lên GitHub**
```bash
git push origin main
```
→ `origin` là tên mặc định của repository trên GitHub, `main` là tên nhánh (branch) chính.

**4. `git pull` — Lấy thay đổi mới nhất về**
```bash
git pull origin main
```
→ Rất quan trọng khi làm việc nhóm: luôn `pull` trước khi bắt đầu code để tránh xung đột (conflict).

### Ví dụ quy trình hoàn chỉnh

```bash

# Bước 4: Commit lại thay đổi
git commit -m "Thêm bài tập vòng lặp for"

# Bước 5: Đẩy lên GitHub
git push origin main

# Bước 6 (lần sau): Lấy code mới nhất trước khi làm tiếp
git pull origin main
```
