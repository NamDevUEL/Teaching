# 1️⃣ GIỚI THIỆU PYTHON

## 📖 Theory (Lý thuyết)

### Python là gì?

**Python** là một ngôn ngữ lập trình bậc cao, được tạo ra bởi **Guido van Rossum** vào năm **1991**. Python nổi tiếng vì:

- ✅ **Cú pháp đơn giản, gần với tiếng Anh tự nhiên** — dễ học, dễ đọc
- ✅ **Đa năng**: Web, AI, Machine Learning, Khoa học dữ liệu, Tự động hóa, Game...
- ✅ **Miễn phí và mã nguồn mở**
- ✅ **Cộng đồng cực lớn**, có hàng ngàn thư viện hỗ trợ (NumPy, Pandas, TensorFlow...)

### Vì sao Python quan trọng trong Khoa học Dữ liệu?

```
┌──────────────────────────────────────────────┐
│              PYTHON ECOSYSTEM                 │
│                                                │
│   📊 Pandas      → Xử lý dữ liệu bảng          │
│   🔢 NumPy       → Tính toán số học            │
│   📈 Matplotlib  → Trực quan hóa dữ liệu       │

│   🧠 TensorFlow  → Deep Learning               │
└──────────────────────────────────────────────┘
```

### So sánh Python với các ngôn ngữ khác

| Tiêu chí | Python | Java | C++ |

# 2️⃣ CÀI ĐẶT PYTHON

## 📖 Theory

Để lập trình Python trên máy tính cá nhân, bạn cần cài đặt **trình thông dịch Python** (Python Interpreter).

### Các bước cài đặt Python trên Windows/macOS

```
BƯỚC 1: Truy cập https://www.python.org/downloads/
              │
              ▼
BƯỚC 2: Tải phiên bản Python mới nhất (khuyến nghị 3.10+)
              │
              ▼
BƯỚC 3: Chạy file cài đặt
         ⚠️ QUAN TRỌNG: Tick chọn
         ☑ "Add Python to PATH"
              │
              ▼
BƯỚC 4: Nhấn "Install Now" và chờ hoàn tất
              │
              ▼
BƯỚC 5: Mở Terminal / Command Prompt, gõ lệnh kiểm tra
```

### Kiểm tra cài đặt thành công

Mở **Terminal** (macOS/Linux) hoặc **Command Prompt** (Windows) và gõ:

```bash
python --version
```

Nếu thấy kết quả dạng `Python 3.x.x` → cài đặt **thành công** ✅

### Lưu ý thường gặp

| Vấn đề | Nguyên nhân | Cách khắc phục |
|---|---|---|
| `'python' is not recognized` | Chưa tick "Add to PATH" | Cài lại và tick chọn |
| Có nhiều phiên bản Python | Cài nhiều lần | Dùng `python3 --version` |
| macOS báo lỗi quyền | Thiếu quyền admin | Dùng `sudo` khi cài |

# 4️⃣ JUPYTER NOTEBOOK

## 📖 Theory

**Jupyter Notebook** là công cụ cho phép viết code Python xen kẽ với văn bản giải thích (Markdown), hình ảnh, biểu đồ... trong cùng một file `.ipynb`.

### Vì sao Jupyter Notebook rất phổ biến trong Data Science?

- 📊 Chạy từng **ô lệnh (cell)** riêng biệt → dễ debug, thử nghiệm
- 📝 Có thể viết giải thích bằng Markdown ngay bên cạnh code
- 📈 Hiển thị biểu đồ, bảng dữ liệu ngay trong notebook
- 🔄 Rất phù hợp cho việc **phân tích dữ liệu từng bước**

### Hai loại Cell chính

```
┌─────────────────────────────────────────┐
│  📝 MARKDOWN CELL                        │
│  Dùng để viết văn bản, tiêu đề, giải thích│
│  Ví dụ: # Đây là tiêu đề                 │
└─────────────────────────────────────────┘

┌─────────────────────────────────────────┐
│  💻 CODE CELL                            │
│  Dùng để viết và chạy code Python         │
│  Ví dụ: print("Hello")                    │
└─────────────────────────────────────────┘
```

### Cài đặt Jupyter Notebook

```bash
pip install notebook
jupyter notebook
```

### Các phím tắt quan trọng

| Phím tắt | Chức năng |
|---|---|
| `Shift + Enter` | Chạy cell hiện tại, nhảy sang cell kế |
| `Ctrl + Enter` | Chạy cell hiện tại, giữ nguyên vị trí |
| `A` | Thêm cell phía trên (chế độ Command) |
| `B` | Thêm cell phía dưới (chế độ Command) |
| `D D` (nhấn D 2 lần) | Xóa cell |
| `M` | Chuyển cell thành Markdown |
| `Y` | Chuyển cell thành Code |

## 💻 Example

Đây chính là một Code Cell — hãy nhấn **Shift + Enter** để chạy nó:

# 7️⃣ DATA TYPES (KIỂU DỮ LIỆU)

## 📖 Theory

Trong Python, mỗi giá trị đều có một **kiểu dữ liệu** (data type). Các kiểu dữ liệu cơ bản gồm:

| Kiểu dữ liệu | Tên tiếng Anh | Ví dụ | Mô tả |
|---|---|---|---|
| Số nguyên | `int` | `5`, `-10`, `100` | Số không có phần thập phân |
| Số thực | `float` | `3.14`, `-0.5` | Số có phần thập phân |
| Chuỗi | `str` | `"Xin chào"` | Văn bản, đặt trong dấu `""` hoặc `''` |
| Boolean | `bool` | `True`, `False` | Giá trị đúng/sai |

### Sơ đồ phân loại

```
                    DATA TYPES
                        │
        ┌───────────────┼───────────────┬──────────────┐
        ▼               ▼               ▼              ▼
      int             float            str            bool

     5, -3            3.14         "Hello"        True, False
```

### Kiểm tra kiểu dữ liệu

````python
# Chuyển đổi kiểu dữ liệu (Type Conversion)
so_string = "100"          # chuỗi
so_int = int(so_string)    # chuyển thành số nguyên
print(so_int + 1)          # 101

so_int2 = 50
so_float = float(so_int2)  # chuyển int thành float
print(so_float)            # 50.0

so_thanh_chuoi = str(123)  # chuyển số thành chuỗi
print("Số " + so_thanh_chuoi + " đã thành chuỗi")
````

# 🔟 OPERATORS (TOÁN TỬ)

## 📖 Theory

### 1. Toán tử số học (Arithmetic Operators)

| Toán tử | Ý nghĩa | Ví dụ | Kết quả |
|---|---|---|---|
| `+` | Cộng | `5 + 3` | `8` |
| `-` | Trừ | `5 - 3` | `2` |
| `*` | Nhân | `5 * 3` | `15` |
| `/` | Chia (kết quả là float) | `5 / 2` | `2.5` |
| `//` | Chia lấy phần nguyên | `5 // 2` | `2` |
| `%` | Chia lấy phần dư (modulo) | `5 % 2` | `1` |
| `**` | Lũy thừa | `5 ** 2` | `25` |

### 2. Toán tử so sánh (Comparison Operators)

| Toán tử | Ý nghĩa | Ví dụ | Kết quả |
|---|---|---|---|
| `==` | Bằng | `5 == 5` | `True` |
| `!=` | Khác | `5 != 3` | `True` |
| `>` | Lớn hơn | `5 > 3` | `True` |
| `<` | Nhỏ hơn | `5 < 3` | `False` |
| `>=` | Lớn hơn hoặc bằng | `5 >= 5` | `True` |
| `<=` | Nhỏ hơn hoặc bằng | `5 <= 4` | `False` |

### 3. Toán tử logic (Logical Operators)

| Toán tử | Ý nghĩa | Ví dụ |
|---|---|---|
| `and` | Đúng khi cả hai đều đúng | `True and False` → `False` |
| `or` | Đúng khi ít nhất một đúng | `True or False` → `True` |
| `not` | Phủ định | `not True` → `False` |

## 💻 Example


````python
# Toán tử số học
a, b = 10, 3
print("Cộng:", a + b)
print("Trừ:", a - b)
print("Nhân:", a * b)
print("Chia:", a / b)
print("Chia lấy nguyên:", a // b)
print("Chia lấy dư:", a % b)
print("Lũy thừa:", a ** 2)
````

````python
# Toán tử so sánh
x, y = 5, 8
print(x == y)   # False
print(x != y)   # True
print(x < y)    # True

# Toán tử logic
diem = 8
tuoi = 20
print(diem >= 5 and tuoi >= 18)   # True and True -> True
print(diem >= 9 or tuoi >= 18)    # False or True -> True
print(not (diem >= 5))            # not True -> False
````

# 1️⃣1️⃣ LIST (DANH SÁCH)

## 📖 Theory

**List** là một cấu trúc dữ liệu dùng để lưu trữ **nhiều giá trị** trong **một biến duy nhất**, có thứ tự và **có thể thay đổi** (mutable).

### Cú pháp khai báo List

```python
danh_sach = [phan_tu_1, phan_tu_2, phan_tu_3]
```

### Sơ đồ minh họa Index (chỉ số)

```
   danh_sach = ["táo", "cam", "chuối", "nho"]

   Index:        0        1         2        3
              ┌───────┬────────┬─────────┬───────┐
              │ "táo" │ "cam"  │"chuối"  │ "nho" │
              └───────┴────────┴─────────┴───────┘
   Index âm:    -4       -3        -2       -1
```

⚠️ Python đánh **chỉ số (index) bắt đầu từ 0**, không phải từ 1!

### Các thao tác phổ biến với List

| Thao tác | Cú pháp | Ví dụ |
|---|---|---|
| Truy cập phần tử | `list[index]` | `danh_sach[0]` |
| Thêm phần tử vào cuối | `list.append(x)` | `danh_sach.append("dâu")` |
| Xóa phần tử | `list.remove(x)` | `danh_sach.remove("cam")` |
| Độ dài danh sách | `len(list)` | `len(danh_sach)` |
| Cắt danh sách (slicing) | `list[start:end]` | `danh_sach[0:2]` |
| Sắp xếp | `list.sort()` | `danh_sach.sort()` |

## 💻 Example


````python
# Khai báo List
trai_cay = ["táo", "cam", "chuối", "nho"]
print(trai_cay)

# Truy cập phần tử theo index
print("Phần tử đầu tiên:", trai_cay[0])
print("Phần tử cuối cùng:", trai_cay[-1])

# Độ dài list
print("Số lượng phần tử:", len(trai_cay))
````

````python
# Thêm và xóa phần tử
trai_cay = ["táo", "cam", "chuối"]

trai_cay.append("dâu")         # thêm vào cuối
print("Sau khi thêm:", trai_cay)

trai_cay.remove("cam")         # xóa phần tử "cam"
print("Sau khi xóa:", trai_cay)

# Slicing - cắt danh sách
so = [10, 20, 30, 40, 50]
print("2 phần tử đầu:", so[0:2])
print("3 phần tử cuối:", so[2:])

# Sắp xếp
diem = [8, 5, 9, 3, 7]
diem.sort()
print("Sau khi sắp xếp:", diem)
````

# 1️⃣2️⃣ TUPLE (BỘ DỮ LIỆU)

## 📖 Theory

**Tuple** giống như List (có thứ tự, có index), nhưng khác biệt lớn nhất là **KHÔNG THỂ thay đổi** (immutable) sau khi tạo.

### Cú pháp khai báo Tuple

```python
bien = (phan_tu_1, phan_tu_2, phan_tu_3)
```

### So sánh List và Tuple

| Đặc điểm | List | Tuple |
|---|---|---|
| Ký hiệu | `[ ]` | `( )` |
| Có thể thay đổi? | ✅ Có (mutable) | ❌ Không (immutable) |
| Tốc độ xử lý | Chậm hơn | **Nhanh hơn** |
| Dùng khi nào | Dữ liệu cần thay đổi | Dữ liệu **cố định** (VD: tọa độ, ngày tháng) |

### Khi nào dùng Tuple?

```
Ví dụ thực tế: Tọa độ địa lý của Cần Thơ (không đổi)
      toa_do = (10.0452, 105.7469)   # (vĩ độ, kinh độ)
```

## 💻 Example


````python
# Khai báo Tuple
toa_do = (10.0452, 105.7469)
print(toa_do)
print("Vĩ độ:", toa_do[0])
print("Kinh độ:", toa_do[1])

# Tuple không thể thay đổi -> dòng dưới đây sẽ báo lỗi nếu bỏ comment
# toa_do[0] = 99  # TypeError: 'tuple' object does not support item assignment
````

````python
# Unpacking Tuple (giải nén giá trị)
thong_tin_sv = ("Lan", 19, "Khoa học Dữ liệu")
ten, tuoi, nganh = thong_tin_sv
print("Tên:", ten)
print("Tuổi:", tuoi)
print("Ngành:", nganh)
````

# 1️⃣3️⃣ DICTIONARY (TỪ ĐIỂN)

## 📖 Theory

**Dictionary** lưu trữ dữ liệu theo cặp **key (khóa) - value (giá trị)**, giống như một cuốn từ điển: tra "khóa" để lấy ra "nghĩa".

### Cú pháp khai báo Dictionary

```python
bien = {
    "key1": value1,
    "key2": value2
}
```

### Sơ đồ minh họa

```
   sinh_vien = {
        "ten":  "Lan",        ──►  key "ten"     → value "Lan"
        "tuoi": 19,            ──►  key "tuoi"    → value 19
        "nganh": "Data Science"──►  key "nganh"   → value "Data Science"
   }
```

### Các thao tác phổ biến

| Thao tác | Cú pháp | Ví dụ |
|---|---|---|
| Truy cập giá trị | `dict[key]` | `sinh_vien["ten"]` |
| Thêm/cập nhật | `dict[key] = value` | `sinh_vien["tuoi"] = 20` |
| Xóa | `del dict[key]` | `del sinh_vien["tuoi"]` |
| Lấy tất cả key | `dict.keys()` | |
| Lấy tất cả value | `dict.values()` | |
| Kiểm tra key tồn tại | `key in dict` | `"ten" in sinh_vien` |

## 💻 Example


````python
# Khai báo Dictionary
sinh_vien = {
    "ten": "Lan",
    "tuoi": 19,
    "nganh": "Khoa học Dữ liệu"
}

print(sinh_vien)
print("Tên sinh viên:", sinh_vien["ten"])
print("Ngành học:", sinh_vien["nganh"])
````

````python
# Thêm, cập nhật, xóa dữ liệu trong Dictionary
sinh_vien = {"ten": "Lan", "tuoi": 19}

sinh_vien["truong"] = "Đại học Cần Thơ"   # thêm mới
print("Sau khi thêm:", sinh_vien)

sinh_vien["tuoi"] = 20                     # cập nhật giá trị
print("Sau khi cập nhật tuổi:", sinh_vien)

del sinh_vien["truong"]                    # xóa key
print("Sau khi xóa 'truong':", sinh_vien)

# Duyệt qua Dictionary
for key, value in sinh_vien.items():
    print(key, ":", value)
````

# 1️⃣5️⃣ INTRODUCTION TO GIT & GITHUB

## 📖 Theory

### Git là gì?

**Git** là một hệ thống **quản lý phiên bản (Version Control System)**, giúp bạn theo dõi lịch sử thay đổi của code theo thời gian, và dễ dàng khôi phục lại phiên bản cũ khi cần.

```
┌─────────────────────────────────────────────────────────┐
│                    LỊCH SỬ CODE VỚI GIT                  │
│                                                           │
│  commit 1 ──► commit 2 ──► commit 3 ──► commit 4 (hiện tại)│
│  "Tạo file"   "Sửa lỗi"    "Thêm tính   "Hoàn thiện"      │
│                             năng mới"                     │
└─────────────────────────────────────────────────────────┘
```

### GitHub là gì?

**GitHub** là một nền tảng **lưu trữ code trực tuyến** dựa trên Git, cho phép:
- 💾 Lưu trữ code trên "đám mây"
- 👥 Cộng tác làm việc nhóm
- 📁 Quản lý các phiên bản dự án (Repository)
- 🌐 Xây dựng portfolio cá nhân

### Vì sao Data Scientist cần GitHub?

```
┌───────────────────────────────────────────────────┐
│         LÝ DO DATA SCIENTIST CẦN GITHUB            │
├───────────────────────────────────────────────────┤
│  📊 Lưu trữ notebook phân tích dữ liệu              │
│  🤝 Cộng tác với team trong dự án AI/DS             │
│  📈 Xây dựng Portfolio cá nhân (nhà tuyển dụng xem) │
│  🔄 Theo dõi phiên bản của Model/Code               │
│  🌍 Tham gia dự án mã nguồn mở (Open Source)        │
└───────────────────────────────────────────────────┘
```

### Sơ đồ khái niệm Git vs GitHub

```
   ┌─────────────┐         push          ┌──────────────┐
   │   Máy tính  │  ───────────────────► │    GitHub     │
   │   cá nhân   │                       │  (trên Cloud) │
   │  (Local Git)│  ◄─────────────────── │               │
   └─────────────┘         pull          └──────────────┘
```

## 💻 Example — Quy trình tạo GitHub và Upload file

### Bước 1: Tạo tài khoản GitHub

```
1. Truy cập https://github.com
2. Nhấn "Sign up"
3. Điền: Email → Password → Username
4. Xác thực email
5. Hoàn tất! Bạn đã có tài khoản GitHub 🎉
```

### Bước 2: Tạo Repository đầu tiên

```
┌──────────────────────────────────────────────┐
│  1. Nhấn nút "+" góc trên phải → "New repository" │
├──────────────────────────────────────────────┤
│  2. Đặt tên Repository (VD: "python-chapter1") │
├──────────────────────────────────────────────┤
│  3. Chọn "Public" (công khai) hoặc "Private"   │
├──────────────────────────────────────────────┤
│  4. Tick chọn "Add a README file"              │
├──────────────────────────────────────────────┤
│  5. Nhấn "Create repository"                   │
└──────────────────────────────────────────────┘
```

### Bước 3: Upload file Python (.py) hoặc Notebook (.ipynb)

**Cách 1 — Upload trực tiếp trên Web (đơn giản nhất cho người mới):**

```
1. Vào Repository vừa tạo
2. Nhấn "Add file" → "Upload files"
3. Kéo thả file .py hoặc .ipynb vào
4. Viết mô tả ngắn (commit message), VD: "Thêm bài tập chương 1"
5. Nhấn "Commit changes"
```

**Cách 2 — Dùng Git command line:**

```bash
git init                                   # khởi tạo Git trong thư mục
git add hello.py                           # thêm file vào vùng theo dõi
git commit -m "Thêm chương trình Hello World"  # lưu lại thay đổi (commit)
git remote add origin <link-repository>    # liên kết với GitHub
git push origin main                       # đẩy code lên GitHub
```

### README.md là gì?

**README.md** là file mô tả (viết bằng Markdown) hiển thị **ngay trang chủ** của Repository, giúp người khác hiểu:
- Dự án này làm gì
- Cách cài đặt / sử dụng
- Tác giả là ai

```
┌───────────────────────────────────┐
│   📁 my-repository                │
│   ├── 📄 README.md   ← Hiển thị   │
│   │      đầu tiên khi ai đó       │
│   │      ghé thăm Repository      │
│   ├── 📄 hello.py                 │
│   └── 📄 chapter1.ipynb           │
└───────────────────────────────────┘
```

**Ví dụ nội dung một file README.md đơn giản:**

```markdown
# Python Chapter 1

Đây là repository chứa bài tập Chương 1 môn Python.

## Nội dung
- hello.py: chương trình Hello World
- chapter1.ipynb: notebook bài giảng đầy đủ

## Tác giả
Nguyễn Văn A - Sinh viên năm nhất
```
