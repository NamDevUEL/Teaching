# 1️⃣ SERIES

## 📖 Theory

**Series** là cấu trúc dữ liệu **1 chiều** của Pandas — giống như một cột dữ liệu (hoặc List "nâng cấp"), mỗi giá trị đều có một **nhãn (index)** đi kèm.

### Cài đặt & Import Pandas

```bash
pip install pandas
```

```python
import pandas as pd    # "pd" là quy ước viết tắt phổ biến
```

### So sánh List và Series

| Đặc điểm | List | Pandas Series |
|---|---|---|
| Có index tùy chỉnh? | Không (chỉ 0,1,2...) | **Có** — có thể đặt tên |
| Tính toán vector hóa | Không | **Có** (giống NumPy) |
| Xử lý dữ liệu thiếu | Khó | Dễ dàng |

### Sơ đồ minh họa Series

```
        Index          Value
      ┌────────┬────────────────┐
      │   0     │      8.5        │
      │   1     │      7.0        │
      │   2     │      9.2        │
      │   3     │      6.5        │
      └────────┴────────────────┘
            (giống 1 CỘT trong bảng Excel)
```

## 💻 Example


````python
import pandas as pd

# Tạo Series từ list
diem = pd.Series([8.5, 7.0, 9.2, 6.5])
print(diem)
print("\nKiểu dữ liệu:", type(diem))
````

````python
# Tạo Series với index tùy chỉnh (tên sinh viên làm nhãn)
diem = pd.Series([8.5, 7.0, 9.2, 6.5], index=["An", "Bình", "Chi", "Dũng"])
print(diem)

print("\nĐiểm của Chi:", diem["Chi"])
print("Điểm trung bình:", diem.mean())
print("Điểm cao nhất:", diem.max())
````

# 2️⃣ DATAFRAME

## 📖 Theory

**DataFrame** là cấu trúc dữ liệu **2 chiều** (dạng bảng) của Pandas — gồm nhiều **dòng (row)** và **cột (column)**, giống hệt một bảng tính Excel. Đây là cấu trúc dữ liệu **quan trọng nhất** trong Pandas.

### Cách tạo DataFrame

```python
df = pd.DataFrame({
    "cot_1": [gia_tri_1, gia_tri_2, ...],
    "cot_2": [gia_tri_1, gia_tri_2, ...],
})
```

### Sơ đồ cấu trúc DataFrame

```
                cột "ten"   cột "tuoi"   cột "diem"
              ┌───────────┬────────────┬───────────┐
   dòng 0     │   "An"     │    19       │    8.5     │
   dòng 1     │   "Bình"   │    20       │    7.0     │
   dòng 2     │   "Chi"    │    19       │    9.2     │
              └───────────┴────────────┴───────────┘
                    ▲
              Mỗi CỘT chính là 1 Series
```

## 💻 Example


````python
import pandas as pd

# Tạo DataFrame mẫu: dữ liệu sinh viên
data_sv = {
    "ten": ["An", "Bình", "Chi", "Dũng", "Em"],
    "tuoi": [19, 20, 19, 21, 20],
    "nganh": ["Data Science", "CNTT", "Data Science", "Kinh tế", "CNTT"],
    "diem_tb": [8.5, 7.0, 9.2, 6.5, 8.0]
}

df = pd.DataFrame(data_sv)
df   # trong Jupyter, gõ tên biến DataFrame sẽ hiển thị dạng bảng đẹp
````

````python
# Các thuộc tính hữu ích để khám phá DataFrame
print("Kích thước (dòng, cột):", df.shape)
print("\nTên các cột:", list(df.columns))
print("\nThông tin tổng quan:")
df.info()
````

````python
# Xem nhanh dữ liệu
print("5 dòng đầu tiên:")
print(df.head())      # mặc định 5 dòng đầu

print("\nThống kê mô tả (chỉ áp dụng cho cột số):")
print(df.describe())
````

# 3️⃣ READ CSV

## 📖 Theory

**CSV** (Comma-Separated Values) là định dạng file phổ biến nhất để lưu trữ dữ liệu dạng bảng — mỗi dòng là 1 bản ghi, các giá trị cách nhau bởi dấu phẩy.

### Cú pháp đọc file CSV

```python
df = pd.read_csv("duong_dan_file.csv")
```

### Một số tham số hữu ích của `read_csv()`

| Tham số | Ý nghĩa |
|---|---|
| `sep=";"` | Chỉ định dấu phân cách khác (mặc định là dấu phẩy) |
| `encoding="utf-8"` | Mã hóa ký tự (quan trọng với tiếng Việt có dấu) |
| `header=0` | Dòng nào chứa tên cột (mặc định dòng đầu tiên) |
| `usecols=["a","b"]` | Chỉ đọc một số cột nhất định |

## 💻 Example

> 💡 Vì notebook này không có sẵn file CSV, ta sẽ **tự tạo** một file CSV mẫu trước, sau đó đọc lại bằng `read_csv()` — quy trình y hệt khi làm việc với file thật.

# 5️⃣ SELECTION (CHỌN DỮ LIỆU)

## 📖 Theory

Pandas cung cấp nhiều cách để **chọn** cột, dòng, hoặc một vùng dữ liệu cụ thể trong DataFrame.

### Các cách chọn dữ liệu phổ biến

| Cú pháp | Ý nghĩa |
|---|---|
| `df["ten_cot"]` | Chọn 1 cột (trả về Series) |
| `df[["cot1","cot2"]]` | Chọn nhiều cột (trả về DataFrame) |
| `df.loc[index]` | Chọn theo **nhãn (label)** của dòng/cột |
| `df.iloc[vi_tri]` | Chọn theo **vị trí số** (giống index của List) |
| `df.loc[dong, cot]` | Chọn 1 giá trị cụ thể theo nhãn |

### Sơ đồ minh họa `loc` vs `iloc`

```
        index    ten      tuoi     diem
          0      "An"      19       8.5
          1      "Bình"    20       7.0
          2      "Chi"     19       9.2

   df.loc[1, "ten"]   → dùng NHÃN (index=1, cột "ten")   → "Bình"
   df.iloc[1, 0]      → dùng VỊ TRÍ (dòng 1, cột 0)       → "Bình"
```

## 💻 Example


````python
import pandas as pd

df = pd.DataFrame({
    "ten": ["An", "Bình", "Chi", "Dũng", "Em"],
    "tuoi": [19, 20, 19, 21, 20],
    "nganh": ["Data Science", "CNTT", "Data Science", "Kinh tế", "CNTT"],
    "diem_tb": [8.5, 7.0, 9.2, 6.5, 8.0]
})

# Chọn 1 cột -> trả về Series
print("Cột 'ten':\n", df["ten"])

# Chọn nhiều cột -> trả về DataFrame
print("\nCột 'ten' và 'diem_tb':\n", df[["ten", "diem_tb"]])
````

````python
# Chọn theo dòng bằng .loc (nhãn) và .iloc (vị trí)
print("Dòng có index = 2 (loc):\n", df.loc[2])

print("\nDòng ở vị trí thứ 0 (iloc):\n", df.iloc[0])

# Chọn 1 giá trị cụ thể
print("\nTên sinh viên ở dòng 3:", df.loc[3, "ten"])

# Chọn nhiều dòng và cột cùng lúc
print("\n2 dòng đầu, 2 cột đầu:\n", df.iloc[0:2, 0:2])
````

# 6️⃣ FILTERING (LỌC DỮ LIỆU)

## 📖 Theory

**Filtering** trong Pandas dùng để **lọc ra các dòng thỏa mãn điều kiện** — tương tự `filter()` ở Chương 3, nhưng áp dụng trực tiếp trên DataFrame.

### Cú pháp lọc dữ liệu

```python
df[dieu_kien]
```

### Kết hợp nhiều điều kiện

| Toán tử | Ý nghĩa | Lưu ý |
|---|---|---|
| `&` | và (and) | Phải đặt mỗi điều kiện trong `()` |
| `\|` | hoặc (or) | Phải đặt mỗi điều kiện trong `()` |
| `~` | phủ định (not) | |

```python
df[(dieu_kien_1) & (dieu_kien_2)]     # ĐÚNG cú pháp
df[dieu_kien_1 & dieu_kien_2]         # ❌ có thể lỗi nếu thiếu ngoặc
```

## 💻 Example


````python
# Lọc sinh viên có điểm trung bình >= 8
sv_gioi = df[df["diem_tb"] >= 8]
print("Sinh viên giỏi (điểm >= 8):")
print(sv_gioi)
````

````python
# Kết hợp nhiều điều kiện: sinh viên ngành Data Science VÀ điểm >= 8
sv_ds_gioi = df[(df["nganh"] == "Data Science") & (df["diem_tb"] >= 8)]
print("Sinh viên Data Science có điểm >= 8:")
print(sv_ds_gioi)

# Điều kiện OR: sinh viên ngành CNTT HOẶC Kinh tế
sv_cntt_kt = df[(df["nganh"] == "CNTT") | (df["nganh"] == "Kinh tế")]
print("\nSinh viên ngành CNTT hoặc Kinh tế:")
print(sv_cntt_kt)
````

````python
# Lọc bằng phương thức .isin() - kiểm tra giá trị có nằm trong 1 danh sách không
sv_theo_nganh = df[df["nganh"].isin(["Data Science", "Kinh tế"])]
print(sv_theo_nganh)
````

# 7️⃣ MISSING VALUES (DỮ LIỆU THIẾU)

## 📖 Theory

Trong thực tế, dữ liệu **hiếm khi hoàn hảo** — thường có ô bị trống (Missing Values), ký hiệu là `NaN` (Not a Number) trong Pandas. Xử lý dữ liệu thiếu là bước **bắt buộc** trước khi phân tích.

### Các thao tác xử lý dữ liệu thiếu

| Hàm | Ý nghĩa |
|---|---|
| `df.isnull()` | Kiểm tra ô nào bị thiếu (`True`/`False`) |
| `df.isnull().sum()` | Đếm số lượng giá trị thiếu theo từng cột |
| `df.dropna()` | **Xóa** các dòng có giá trị thiếu |
| `df.fillna(gia_tri)` | **Điền** giá trị thay thế vào ô thiếu |

### Sơ đồ minh họa

```
   ten      diem
   "An"      8.5
   "Bình"    NaN     ← dữ liệu thiếu
   "Chi"     9.2
   "Dũng"    NaN     ← dữ liệu thiếu

   dropna()  →  chỉ giữ lại "An" và "Chi" (xóa dòng có NaN)
   fillna(0) →  NaN được thay bằng 0 (giữ nguyên số dòng)
```

## 💻 Example

# 8️⃣ GROUPBY

## 📖 Theory

`groupby()` dùng để **nhóm dữ liệu** theo một hoặc nhiều cột, sau đó áp dụng phép **tổng hợp (aggregation)** như tính tổng, trung bình, đếm số lượng... cho từng nhóm — giống chức năng "Pivot Table" trong Excel.

### Cú pháp

```python
df.groupby("ten_cot")["cot_can_tinh"].ham_tong_hop()
```

### Các hàm tổng hợp phổ biến

| Hàm | Ý nghĩa |
|---|---|
| `.mean()` | Trung bình |
| `.sum()` | Tổng |
| `.count()` | Đếm số lượng |
| `.max()` / `.min()` | Lớn nhất / Nhỏ nhất |

### Sơ đồ minh họa GroupBy

```
   Dữ liệu gốc:                     Nhóm theo "nganh":
   ten    nganh          diem       nganh          diem_trung_binh
   An     Data Science    8.5       Data Science         8.85
   Bình   CNTT            7.0  ──►  CNTT                 7.25
   Chi    Data Science    9.2       Kinh tế              6.5
   Dũng   Kinh tế         6.5
   Em     CNTT            7.5
```

## 💻 Example


````python
import pandas as pd

df = pd.DataFrame({
    "ten": ["An", "Bình", "Chi", "Dũng", "Em", "Phúc"],
    "nganh": ["Data Science", "CNTT", "Data Science", "Kinh tế", "CNTT", "Data Science"],
    "diem_tb": [8.5, 7.0, 9.2, 6.5, 7.5, 8.9]
})

# Nhóm theo ngành, tính điểm trung bình mỗi ngành
diem_theo_nganh = df.groupby("nganh")["diem_tb"].mean()
print(diem_theo_nganh)
````

````python
# Nhóm theo ngành, tính nhiều thống kê cùng lúc bằng .agg()
thong_ke = df.groupby("nganh")["diem_tb"].agg(["mean", "max", "min", "count"])
print(thong_ke)
````

# 9️⃣ MERGE

## 📖 Theory

`merge()` dùng để **kết hợp 2 DataFrame** lại với nhau dựa trên một **cột chung (khóa - key)** — giống thao tác `JOIN` trong SQL.

### Cú pháp

```python
pd.merge(df1, df2, on="cot_chung", how="inner")
```

### Các kiểu merge (`how`)

| Kiểu | Ý nghĩa |
|---|---|
| `inner` (mặc định) | Chỉ giữ dòng có key khớp ở **cả 2** bảng |
| `left` | Giữ **toàn bộ** dòng của bảng trái, khớp được thì điền, không thì `NaN` |
| `right` | Giữ toàn bộ dòng của bảng phải |
| `outer` | Giữ **toàn bộ** dòng của cả 2 bảng |

### Sơ đồ minh họa Merge (kiểu inner)

```
   df_sinh_vien              df_diem                    Kết quả merge
   ┌────┬───────┐            ┌────┬───────┐              ┌────┬───────┬───────┐
   │mssv│ ten    │            │mssv│ diem   │              │mssv│ ten    │ diem   │
   ├────┼───────┤            ├────┼───────┤              ├────┼───────┼───────┤
   │ 01  │ An     │   merge   │ 01  │ 8.5    │    =====►    │ 01  │ An     │ 8.5    │
   │ 02  │ Bình   │  on="mssv"│ 02  │ 7.0    │              │ 02  │ Bình   │ 7.0    │
   │ 03  │ Chi    │            │ 04  │ 9.0    │              └────┴───────┴───────┘
   └────┴───────┘            └────┴───────┘         (mssv "03" và "04" không khớp
                                                        cả 2 bên nên bị loại - inner)
```

## 💻 Example


````python
import pandas as pd

# Bảng 1: thông tin sinh viên
df_sinh_vien = pd.DataFrame({
    "mssv": ["SV01", "SV02", "SV03", "SV04"],
    "ten": ["An", "Bình", "Chi", "Dũng"]
})

# Bảng 2: điểm số (chỉ có 3 sinh viên có điểm, thiếu SV04, thừa SV05)
df_diem = pd.DataFrame({
    "mssv": ["SV01", "SV02", "SV03", "SV05"],
    "diem_tb": [8.5, 7.0, 9.2, 6.0]
})

print("Bảng sinh viên:\n", df_sinh_vien)
print("\nBảng điểm:\n", df_diem)
````

````python
# Merge kiểu "inner" - chỉ giữ MSSV có ở CẢ 2 bảng
merge_inner = pd.merge(df_sinh_vien, df_diem, on="mssv", how="inner")
print("Merge INNER:\n", merge_inner)

# Merge kiểu "left" - giữ TOÀN BỘ sinh viên, kể cả chưa có điểm
merge_left = pd.merge(df_sinh_vien, df_diem, on="mssv", how="left")
print("\nMerge LEFT:\n", merge_left)

# Merge kiểu "outer" - giữ TOÀN BỘ dữ liệu của cả 2 bảng
merge_outer = pd.merge(df_sinh_vien, df_diem, on="mssv", how="outer")
print("\nMerge OUTER:\n", merge_outer)
````

## 📖 Theory

Khi một dự án Data Science phát triển lớn hơn (nhiều file code, dữ liệu, hình ảnh...), việc tổ chức file **lộn xộn** sẽ gây khó khăn cho bản thân và đồng đội. **Project Structure** là cách tổ chức thư mục **chuẩn hóa**, giúp bất kỳ ai (kể cả nhà tuyển dụng xem GitHub của bạn) cũng hiểu được dự án ngay lập tức.

### Cấu trúc thư mục chuẩn cho dự án Data Science

```
project/
│
├── data/                   ← Chứa toàn bộ dữ liệu của dự án
│   ├── raw/                 ← Dữ liệu thô, chưa xử lý (KHÔNG sửa trực tiếp)
│   └── processed/           ← Dữ liệu đã làm sạch, sẵn sàng phân tích
│
├── notebooks/               ← Chứa các file Jupyter Notebook (.ipynb)
│   ├── 01_kham_pha_du_lieu.ipynb
│   └── 02_phan_tich.ipynb
│
├── images/                  ← Hình ảnh, biểu đồ xuất ra từ quá trình phân tích
│   └── bieu_do_doanh_thu.png
│
├── README.md                ← Mô tả tổng quan về dự án
├── requirements.txt         ← Danh sách thư viện cần cài đặt
└── .gitignore                ← Danh sách file/thư mục KHÔNG đưa lên GitHub
```

### Giải thích chi tiết từng thành phần

| Thành phần | Vai trò |
|---|---|
| **data/** | Tách riêng dữ liệu **thô (raw)** và **đã xử lý (processed)** để không nhầm lẫn, dễ tái tạo lại pipeline xử lý |
| **notebooks/** | Chứa các notebook phân tích, nên đặt tên có **số thứ tự** (01, 02...) để biết trình tự làm việc |
| **images/** | Lưu biểu đồ/hình ảnh xuất ra, tiện dùng lại trong báo cáo hoặc README |
| **README.md** | "Bộ mặt" của dự án — mô tả mục tiêu, cách cài đặt, cách chạy, kết quả chính |
| **requirements.txt** | Liệt kê thư viện & phiên bản cần thiết, giúp người khác cài đặt môi trường bằng 1 lệnh |
| **.gitignore** | Loại trừ các file KHÔNG nên đưa lên GitHub (file rác, dữ liệu quá lớn, thông tin nhạy cảm...) |

### Nội dung mẫu file `requirements.txt`

```
pandas==2.2.0
numpy==1.26.0
matplotlib==3.8.0
openpyxl==3.1.2
```

→ Cài đặt toàn bộ chỉ với 1 lệnh:
```bash
pip install -r requirements.txt
```

### Nội dung mẫu file `.gitignore`

```
# Không đưa dữ liệu thô quá lớn lên GitHub
data/raw/*.csv

# Không đưa file tạm của Jupyter lên GitHub
.ipynb_checkpoints/

# Không đưa môi trường ảo Python lên GitHub
venv/
__pycache__/

# Không đưa thông tin nhạy cảm (API key, mật khẩu...) lên GitHub
.env

## 💻 Example (mô phỏng tạo cấu trúc project bằng Python)


````python
# Mô phỏng lại cấu trúc thư mục chuẩn của 1 project Data Science bằng Python
# (chỉ để MINH HỌA cách tổ chức, không nhất thiết phải chạy lệnh tạo thư mục thật trong notebook)

cau_truc_project = {
    "project/": {
        "data/": ["raw/", "processed/"],
        "notebooks/": ["01_kham_pha_du_lieu.ipynb", "02_phan_tich.ipynb"],
        "images/": ["bieu_do_doanh_thu.png"],
        "README.md": None,
        "requirements.txt": None,
        ".gitignore": None
    }
}

def in_cau_truc(cau_truc, cap=0):
    for ten, con in cau_truc.items():
        print("  " * cap + "├── " + ten)
        if isinstance(con, dict):
            in_cau_truc(con, cap + 1)
        elif isinstance(con, list):
            for item in con:
                print("  " * (cap + 1) + "├── " + item)

in_cau_truc(cau_truc_project)
````
